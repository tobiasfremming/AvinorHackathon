```python
import pandas as pd

# Last inn datasettet
df = pd.read_csv('../data/historical_flights.csv')

# Print antall cancelled fly
cancelled_count = df['cancelled'].sum()
total_flights = len(df)
cancellation_rate = (cancelled_count / total_flights) * 100

print(f"Antall cancelled fly: {cancelled_count:,}")
print(f"Total antall fly: {total_flights:,}")
print(f"Cancellation rate: {cancellation_rate:.2f}%")
```
