**Exploratory Data Analysis**

```python
import pandas as pd
import os

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_path = os.path.join("..", "data")

csv_files = [
    "airportgroups.csv",
    "historical_flights.csv",
    "inference_data_oct2025.csv",
    "preds_mal.csv",
    "schedule_oct2025.csv",
    "training_data.csv"
]

# Dictionary to store DataFrames
dataframes = {}

# Read each CSV into a DataFrame
for file in csv_files:
    file_path = os.path.join(data_path, file)
    df_name = file.replace(".csv", "")
    dataframes[df_name] = pd.read_csv(file_path)
    
    print(f"\n=== {df_name} ===")
    print(dataframes[df_name].head(), "\n")
```

```python


df = dataframes["training_data"]

# 1. Target distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["target"], bins=30, ax=axes[0])
axes[0].set_title("Target Distribution")
sns.histplot(df["target2"], bins=30, ax=axes[1])
axes[1].set_title("Target2 Distribution")
plt.show()

# 2. Target by hour of day
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x="hour", y="target", estimator="mean")
plt.title("Average Target by Hour of Day")
plt.show()

# 3. Target by season
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="feat_season", y="target")
plt.title("Target by Season")
plt.show()

# 4. Schedule features vs. Target
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x="feat_sched_flights_cnt", y="target", alpha=0.3, ax=axes[0])
axes[0].set_title("Flights Count vs Target")
sns.scatterplot(data=df, x="feat_sched_concurrence", y="target", alpha=0.3, ax=axes[1])
axes[1].set_title("Concurrence vs Target")
plt.show()

# 5. Target by airport group
plt.figure(figsize=(12, 5))
sns.barplot(data=df, x="airport_group", y="target", estimator="mean")
plt.title("Average Target by Airport Group")
plt.show()
```
