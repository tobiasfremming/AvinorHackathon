```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

data_path = "../data/"


df_airports = pd.read_csv(data_path + "airportgroups.csv")
print(f"Airport Groups: {df_airports.shape}")

df_hist = pd.read_csv(data_path + "historical_flights.csv")
print(f"Historical Flights: {df_hist.shape}")

df_sched = pd.read_csv(data_path + "schedule_oct2025.csv")
print(f"Schedule Oct 2025: {df_sched.shape}")

df_weather = pd.read_csv(data_path + "weather_data_complete.csv")
print(f"Weather Data: {df_weather.shape}")

df_airport_coords = pd.read_csv(data_path + "airport_coordinates.csv")
print(f"Airport Coordinates: {df_airport_coords.shape}")

df_currency = pd.read_csv(data_path + "nok_eur_quarterly.csv")
print(f"Currency Exchange: {df_currency.shape}")




# print("\nDataset Summary:")
# datasets = {
#     'Airport Groups': df_airports,
#     'Historical Flights': df_hist,
#     'Schedule Oct 2025': df_sched,
    
# }

# for name, df in datasets.items():
#     print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")


```

```python
# Historical Flights

df_processed_hist = df_hist.copy()

std_parsed = pd.to_datetime(df_processed_hist['std'])
sta_parsed= pd.to_datetime(df_processed_hist['sta']) if 'sta' in df_processed_hist.columns else None
    
df_processed_hist['std'] = std_parsed
df_processed_hist['sta'] = sta_parsed


df_processed_hist['service_type'] = df_processed_hist['service_type'].fillna('Q')

# todo: missing atd and ata:
# train a model without these values, predict these values, and train a model with these values
# or completely drop these rows

df_processed_hist = df_processed_hist[df_processed_hist['atd'].notnull() & df_processed_hist['ata'].notnull()]

print(df_processed_hist.isna().sum())
print(df_processed_hist.notna().sum())
```

```python
# Weather

df_processed_weather = df_weather.copy()
df_processed_weather['datetime'] = pd.to_datetime(df_processed_weather['datetime'])
```

```python
# currency_exchange

df_processed_currency = df_currency.copy()
df_processed_currency['quarter_start'] = pd.to_datetime(df_processed_currency['quarter_start'])
df_processed_currency['quarter_end'] = pd.to_datetime(df_processed_currency['quarter_end'])

print(df_processed_currency.isna().sum())
```
